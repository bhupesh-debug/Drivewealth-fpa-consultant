```python
# 01 — Exploratory analysis (SYNTHETIC DATA, fictional NorthPeak Securities)
import pandas as pd
monthly = pd.read_csv('data/synthetic/monthly_close.csv')
print(monthly.shape)
print(monthly.head(10).to_string(index=False))
```

```python
rev = monthly[monthly.category=='revenue'].groupby('month')['actual'].sum()
cost = monthly[monthly.category=='cost'].groupby('month')['actual'].sum()
print('Latest month revenue:', round(float(rev.iloc[-1]), 2))
print('Latest month costs:', round(float(cost.iloc[-1]), 2))
print('Latest month net P&L:', round(float(rev.iloc[-1]-cost.iloc[-1]), 2))
print(rev.describe().to_string())
```

```python
clients = pd.read_csv('data/synthetic/client_monthly_revenue.csv')
print(clients.groupby('segment')['revenue'].sum().to_string())
daily = pd.read_csv('data/synthetic/daily_flash.csv')
print('Daily rows:', len(daily), '| mean daily net P&L:', round(float(daily['net_pnl'].mean()), 2))
```
